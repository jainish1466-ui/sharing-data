```python
data = np.array([[15,18,20,24],
                [16,19,21,23],
                [14,17,22,25],
                [15,18,23,24],
                [16,20,22,26]])
```

```python
k=data.shape[0] ## 0 means number of rows
print("Number of Blocks:",k)
```

```python
b=data.shape[1] ## 1 means number of columns
print("Number of Blocks:",b)
```

```python
grand_mean = np.mean(data)
print("Grand Mean",grand_mean)
```

```python
treatment_mean = np.mean(data,axis=0)
print("Treatment Mean",treatment_mean)
```
